```python
import pandas as pd
import numpy as np
import sklearn

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Scikit-learn version:", sklearn.__version__)
print("Machine Learning environment is ready!")
```

```python

# Step 5: Load the Breast Cancer Wisconsin dataset

import pandas as pd

# The dataset has 30 features, an ID and a diagnosis label
feature_names = [
    f"{measurement}_{property_name}"
    for measurement in ["mean", "se", "worst"]
    for property_name in [
        "radius", "texture", "perimeter", "area",
        "smoothness", "compactness", "concavity",
        "concave_points", "symmetry", "fractal_dimension"
    ]
]

column_names = ["id", "diagnosis"] + feature_names

# Read the downloaded UCI dataset
df = pd.read_csv("wdbc.data", header=None, names=column_names)

# Display basic information
print("Dataset loaded successfully!")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Number of input features:", df.shape[1] - 2)

print("\nDiagnosis distribution:")
print(df["diagnosis"].value_counts())

print("\nFirst five records:")
display(df.head())
```

```python

# Step 6A: Dataset quality checks

print("Dataset shape:", df.shape)

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes.value_counts())

print("\nTarget classes:")
print(df["diagnosis"].value_counts())

print("\nFeature summary:")
display(df.drop(columns=["id", "diagnosis"]).describe())
```

```python

# Step 6B: Prepare features and target

from sklearn.model_selection import train_test_split

# Select the 30 numerical input features
X = df.drop(columns=["id", "diagnosis"])

# Encode diagnosis: Benign = 0, Malignant = 1
y = df["diagnosis"].map({"B": 0, "M": 1})

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Number of features:", X_train.shape[1])

print("\nTraining class distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts().sort_index())

# Save the test data for the final GitHub repository
test_data = X_test.copy()
test_data["diagnosis"] = y_test.map({0: "B", 1: "M"})
test_data.to_csv("test_data.csv", index=False)

print("\ntest_data.csv created successfully!")
```

```python

# Step 7: Logistic Regression

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix
)
import joblib
import os

# Create a folder for trained models
os.makedirs("model", exist_ok=True)

# Scaling is fitted only on training data
lr_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

# Train the model
lr_model.fit(X_train, y_train)

# Predict on the test dataset
lr_predictions = lr_model.predict(X_test)
lr_probabilities = lr_model.predict_proba(X_test)[:, 1]

# Calculate the required metrics
lr_metrics = {
    "Accuracy": accuracy_score(y_test, lr_predictions),
    "AUC": roc_auc_score(y_test, lr_probabilities),
    "Precision": precision_score(y_test, lr_predictions),
    "Recall": recall_score(y_test, lr_predictions),
    "F1": f1_score(y_test, lr_predictions),
    "MCC": matthews_corrcoef(y_test, lr_predictions)
}

print("LOGISTIC REGRESSION RESULTS")
print("-" * 35)

for metric, value in lr_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, lr_predictions))

# Save model with its fitted scaler
joblib.dump(lr_model, "model/logistic_regression.joblib")

print("\nModel saved successfully!")
```

```python

# Step 8: Decision Tree Classifier

from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(
    max_depth=5,
    min_samples_leaf=3,
    random_state=42
)

# Train the model
dt_model.fit(X_train, y_train)

# Predict using the test data
dt_predictions = dt_model.predict(X_test)
dt_probabilities = dt_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
dt_metrics = {
    "Accuracy": accuracy_score(y_test, dt_predictions),
    "AUC": roc_auc_score(y_test, dt_probabilities),
    "Precision": precision_score(y_test, dt_predictions),
    "Recall": recall_score(y_test, dt_predictions),
    "F1": f1_score(y_test, dt_predictions),
    "MCC": matthews_corrcoef(y_test, dt_predictions)
}

print("DECISION TREE RESULTS")
print("-" * 35)

for metric, value in dt_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, dt_predictions))

# Save the model
joblib.dump(dt_model, "model/decision_tree.joblib")

print("\nDecision Tree model saved successfully!")
```

```python

# Step 9: K-Nearest Neighbours (KNN)

from sklearn.neighbors import KNeighborsClassifier

# Create KNN pipeline with feature scaling
knn_model = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)

# Train the model
knn_model.fit(X_train, y_train)

# Predict on test data
knn_predictions = knn_model.predict(X_test)
knn_probabilities = knn_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
knn_metrics = {
    "Accuracy": accuracy_score(y_test, knn_predictions),
    "AUC": roc_auc_score(y_test, knn_probabilities),
    "Precision": precision_score(y_test, knn_predictions),
    "Recall": recall_score(y_test, knn_predictions),
    "F1": f1_score(y_test, knn_predictions),
    "MCC": matthews_corrcoef(y_test, knn_predictions)
}

print("KNN RESULTS")
print("-" * 35)

for metric, value in knn_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, knn_predictions))

# Save the trained KNN pipeline
joblib.dump(knn_model, "model/knn.joblib")

print("\nKNN model saved successfully!")
```

```python

# Step 10: Gaussian Naive Bayes Classifier

from sklearn.naive_bayes import GaussianNB

# Create a pipeline with feature scaling
nb_model = make_pipeline(
    StandardScaler(),
    GaussianNB()
)

# Train the model
nb_model.fit(X_train, y_train)

# Predict on the test dataset
nb_predictions = nb_model.predict(X_test)
nb_probabilities = nb_model.predict_proba(X_test)[:, 1]

# Calculate all six evaluation metrics
nb_metrics = {
    "Accuracy": accuracy_score(y_test, nb_predictions),
    "AUC": roc_auc_score(y_test, nb_probabilities),
    "Precision": precision_score(y_test, nb_predictions),
    "Recall": recall_score(y_test, nb_predictions),
    "F1": f1_score(y_test, nb_predictions),
    "MCC": matthews_corrcoef(y_test, nb_predictions)
}

print("GAUSSIAN NAIVE BAYES RESULTS")
print("-" * 35)

for metric, value in nb_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, nb_predictions))

# Save the trained model and preprocessing pipeline
joblib.dump(nb_model, "model/naive_bayes.joblib")

print("\nNaive Bayes model saved successfully!")
```

```python

# Step 11: Support Vector Machine (SVM)

from sklearn.svm import SVC

# Create SVM pipeline with feature scaling
svm_model = make_pipeline(
    StandardScaler(),
    SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        probability=True,
        random_state=42
    )
)

# Train the model
svm_model.fit(X_train, y_train)

# Predict on the test dataset
svm_predictions = svm_model.predict(X_test)
svm_probabilities = svm_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
svm_metrics = {
    "Accuracy": accuracy_score(y_test, svm_predictions),
    "AUC": roc_auc_score(y_test, svm_probabilities),
    "Precision": precision_score(y_test, svm_predictions),
    "Recall": recall_score(y_test, svm_predictions),
    "F1": f1_score(y_test, svm_predictions),
    "MCC": matthews_corrcoef(y_test, svm_predictions)
}

print("SUPPORT VECTOR MACHINE RESULTS")
print("-" * 35)

for metric, value in svm_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, svm_predictions))

# Save the model with preprocessing
joblib.dump(svm_model, "model/svm.joblib")

print("\nSVM model saved successfully!")
```

```python

# Step 12: Compare all five classification models

import pandas as pd

all_metrics = {
    "Logistic Regression": lr_metrics,
    "Decision Tree": dt_metrics,
    "KNN": knn_metrics,
    "Gaussian Naive Bayes": nb_metrics,
    "Support Vector Machine": svm_metrics
}

# Create comparison table
comparison_df = pd.DataFrame.from_dict(
    all_metrics, orient="index"
)

comparison_df.index.name = "ML Model Name"

print("FINAL MODEL COMPARISON")
display(comparison_df.round(4))

# Identify the model with the highest F1 score
winner = comparison_df["F1"].idxmax()

print("\nBest model based on F1 Score:", winner)
print("Best F1 Score:", round(comparison_df.loc[winner, "F1"], 4))

# Save the comparison results
comparison_df.to_csv("model_comparison.csv")

print("\nmodel_comparison.csv saved successfully!")
```
